```python
# importing libraries
import pandas as pd
import numpy as np
```

```python
# read the ratings dataset
r_cols=['user_id','movie_id','rating','unix_timestamp']
ratings=pd.read_csv('ml-100k/u.data',sep='\t',names=r_cols,encoding='latin-1')
```

```python
# Read the movie Dataset
i_cols=['movie id','movie title','release date','video release date','IMDB URL','unknown','Action','Adventure',
        'Animation','Children\'s','Comedy','Crime','Documentary','Drama','Fantasy','Film-Noir','Horror','Musical',
        'Mystery','Romance','Sci-Fi','Thriller','War','Western']
items=pd.read_csv('ml-100k/u.item',sep='|',names=i_cols,encoding='latin-1')
```

```python
print("Ratings shape:",ratings.shape)
print("Movies shape:",items.shape)
```

```python
genres=['Action','Adventure','Animation','Children\'s','Comedy','Crime','Documentary',
            'Drama','Fantasy','Film-Noir','Horror','Musical','Mystery','Romance','Sci-Fi',
            'Thriller','War','Western']
print("Number of genres:",len(genres))
print("Genres:",genres)
```

```python
genre_movie_count={}
for genre in genres:
    count=items[items[genre]==1].shape[0]
    genre_movie_count[genre]=count
```

```python
genre_movie_count
```

```python
movie_rating_count=ratings.groupby('movie_id').size()
movie_rating_count.head()
```

```python
items['rating_count']=items['movie id'].map(movie_rating_count)
items[['movie id','movie title','rating_count']].head()
```

```python
action_movies=items[items['Action']==1]
action_top5=action_movies.sort_values(by='rating_count',ascending=False).head(5)
action_top5[['movie id','movie title','rating_count']]
```

```python
top_5_movies={}
for genre in genres:
    genre_movies=items[items[genre]==1]
    top_5=genre_movies.sort_values(by='rating_count',ascending=False).head(5)
    top_5_movies[genre]=top_5[['movie id','movie title','rating_count']]
```

```python
top_5_movies
```

```python
action_recommendation=top_5_movies['Action']['movie title'].tolist()
```

```python
action_recommendation
```

```python
genre_recommendations={}
for genre in genres:
    movie_list=top_5_movies[genre]['movie title'].tolist()
    genre_recommendations[genre]=movie_list
```

```python
genre_recommendations
```

```python
def popularity_based(genre,genre_recommendations):
    return genre_recommendations[genre]
```

```python
popularity_based('Action',genre_recommendations)
```

```python
for genre,movies in genre_recommendations.items():
    print("\n", genre)
    for i,movie in enumerate(movies,1):
        print(i,movie)
```
